part A. A pollutant in wastewater is degraded in a batch photoreactor by a first-order reaction,dc/dt=-kc.use the ruler method :C_new =C_old*t.take C0=1.0mol/L,k=0.12min^-1,t=0.5min

1) using for loop,simulate 5 minutes and print a table of time(min),concentration(mol/L) and conversion (%)

```python
C0 = 1.0
k = 0.12
dt = 0.5
total_time = 5.0
steps = int(total_time / dt)

C = C0
print(f"{'Time (min)':>10} {'Concentration (mol/L)':>24} {'Conversion (%)':>18}")
print(f"{0:10.1f} {C:24.4f} {0.0:18.2f}")

for step in range(1, steps + 1):
    C = C - k * C * dt
    time = step * dt
    conversion = (C0 - C) / C0 * 100
    print(f"{time:10.1f} {C:24.4f} {conversion:18.2f}")
```

part A. A pollutant in wastewater is degraded in a batch photoreactor by a first-order reaction,dc/dt=-kc.use the ruler method :C_new =C_old*t.take C0=1.0mol/L,k=0.12min^-1,t=0.5min

2)using math.exp(),calculate the exact value C0e^(-kt) st 5 min and print the absolute error of the eler result.

```python
C0 = 1.0
k = 0.12
dt = 0.5
steps = int(5 / dt)

C_euler = C0
for _ in range(steps):
    C_euler -= k * C_euler * dt

C_exact = C0 * math.exp(-k * 5)
absolute_error = abs(C_euler - C_exact)

print(f"Euler concentration at 5 min: {C_euler:.6f} mol/L")
print(f"Exact concentration at 5 min: {C_exact:.6f} mol/L")
print(f"Absolute error: {absolute_error:.6f} mol/L")
```

part A. A pollutant in wastewater is degraded in a batch photoreactor by a first-order reaction,dc/dt=-kc.use the ruler method :C_new =C_old*t.take C0=1.0mol/L,k=0.12min^-1,t=0.5min

3)using a while lopp ,repeat the simulation until conversion reaches 70% or time reaches 100 min.print the time needed and the final concentratiom

```python
C0 = 1.0
k = 0.12
dt = 0.5

C = C0
time = 0.0
conversion = 0.0

while conversion < 70 and time < 100:
    C = C - k * C * dt
    time += dt
    conversion = (C0 - C) / C0 * 100

print(f"Time needed: {time:.1f} min")
print(f"Final concentration: {C:.4f} mol/L")
```

partA. a plant has n candidate feed components and wants to blend r of them (order does not matter).write a recursive function factorial(n) (remember the base case for 0!)and a function combination(n,r)=n!/(r!(n-r)!)
that uses it . reject invalid input (negative values,r.n).print the number of possible blends for n=8,r=4 and show the message for n=3,r=5

```python
def factorial(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n must be a non-negative integer.")
    if n == 0:  # Base case: 0! = 1
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if not isinstance(n, int) or not isinstance(r, int) or n < 0 or r < 0 or r > n:
        raise ValueError("Invalid input: require integers with 0 <= r <= n.")
    return factorial(n) // (factorial(r) * factorial(n - r))


print("For n=8, r=4:", combination(8, 4))

try:
    print("For n=3, r=5:", combination(3, 5))
except ValueError as error:
    print(error)
```

partB. during a cooling experiment,temperature readings readings are taken at times=fibonacci number*3min.write an interative function fib_iter(n) that returns a list of the first n fibonacci number,and a recursive function fir_rec(k) that returns the k-th fibonacci number (F0=0,F1=1). for n=8,check that both functions agree and print a table: reading no. | fibonacci tag | time(min)

```python
def fib_iter(n):
    numbers = []
    a, b = 0, 1
    for _ in range(n):
        numbers.append(a)
        a, b = b, a + b
    return numbers


def fib_rec(k):
    if k < 0:
        raise ValueError("k must be non-negative.")
    if k < 2:
        return k
    return fib_rec(k - 1) + fib_rec(k - 2)


n = 8
tags = fib_iter(n)

# Check the iterative list against the recursive function.
assert tags == [fib_rec(k) for k in range(n)]

print(f"{'Reading no.':<14}{'Fibonacci tag':<16}{'Time (min)'}")
for reading, tag in enumerate(tags, start=1):
    print(f"{reading:<14}{tag:<16}{tag * 3}")
```

volume_L                                              pressure_bar
 2                                                     3.01

 plot pressure vs volume as a scatter plot.calculate k as mean of p*vand draw the reference curve p=k
 /v on same graph . add title,labels with units and legend

```python
import numpy as np
import matplotlib.pyplot as plt

volume_L = np.array([2, 3, 4, 5, 6])
pressure_bar = np.array([3.01, 1.98, 1.52, 1.19, 1.01])

k = np.mean(volume_L * pressure_bar)
print(f"k = {k:.2f} bar·L")

volume_curve = np.linspace(volume_L.min(), volume_L.max(), 300)
pressure_curve = k / volume_curve

plt.scatter(volume_L, pressure_bar, label="Measured data")
plt.plot(volume_curve, pressure_curve, label=f"Reference curve: P = {k:.2f}/V")

plt.title("Air Compression at Constant Temperature")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
